## Seaborn Library

### Statistical Graphics

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'x': [1, 2, 3, 4, 5],
    'y': [2, 4, 3, 5, 1],
    'group': ['X', 'Y', 'X', 'Y', 'X']
}

df = pd.DataFrame(data)
sns.scatterplot(x='x', y='y', hue='group', data=df)

plt.title('Scatter Plot with Group-Based Hue')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

### Joint Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'x': [1, 2, 3, 4, 5],
    'y': [2, 5, 3, 8, 10]
}

sns.jointplot(x='x', y='y', kind='scatter', data=data)
plt.show()
```

### Pair Plot

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'x': [1, 2, 3, 4, 5],
    'y': [2, 4, 3, 5, 1],
    'z': [10, 15, 20, 25, 30]
}

df = pd.DataFrame(data)
sns.pairplot(df)
plt.show()
```

### Grid Plot

```python
import matplotlib.pyplot as plt

Grid_plot = plt.GridSpec(2, 3, wspace=0.8, hspace=0.6)

plt.subplot(Grid_plot[0, 0])
plt.grid(True)

plt.subplot(Grid_plot[0, 1:])
plt.grid(True)

plt.subplot(Grid_plot[1, :2])
plt.grid(True)

plt.subplot(Grid_plot[1, 2])
plt.grid(True)

plt.show()
```

### Subplots

## Multiple Plots

### FacetGrid – Column

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'x': [1, 2, 3, 4, 5],
    'y': [2, 4, 3, 5, 1],
    'hue': ['A', 'B', 'A', 'B', 'A']
}

df = pd.DataFrame(data)
g = sns.FacetGrid(df, col='hue', height=4)
g.map(sns.scatterplot, 'x', 'y')
plt.show()
```

## Scatter Plot, Line Plot and Histogram

### Scatter Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'x_values': [1, 2, 3, 4, 5],
    'y_values': [2, 5, 3, 8, 10]
}

sns.scatterplot(x='x_values', y='y_values', data=data)
plt.title('Scatterplot Example')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

### Histogram

## Box Plot

### Basic Box Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'A': [1, 2, 3, 4, 5],
    'B': [2, 4, 6, 8, 10],
    'C': [3, 6, 9, 12, 15]
}

sns.boxplot(data=data)
plt.title('Box Plot Example')
plt.xlabel('Columns')
plt.ylabel('Values')
plt.show()
```

### Box Plot with Outliers

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'A': [1, 2, 3, 4, 5, 30],
    'B': [2, 4, 6, 8, 7, 28],
    'C': [3, 6, 9, 5, 2, 7]
}

df = pd.DataFrame(data)
sns.boxplot(data=df, showmeans=True, whis=1.5)
plt.title('Box Plot with Outliers')
plt.xlabel('Columns')
plt.ylabel('Values')
plt.show()
```

## Playing with Text

### Word Cloud

```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud

text = "This is a sample text for creating a word cloud. Word clouds are a great way to visualize the frequency of words in a text."

wordcloud = WordCloud(width=800, height=400, background_color='white')
wordcloud.generate(text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.show()
```

### Text Annotation on a Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 5, 1]

sns.scatterplot(x=x, y=y)

for i in range(len(x)):
    plt.text(x[i], y[i], f'({x[i]}, {y[i]})', ha='center', va='bottom')

plt.show()
```

### Text as a Plot Element

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

ax.text(0.5, 0.5, 'This is a sample text on the plot',
        horizontalalignment='center',
        verticalalignment='center',
        transform=ax.transAxes)

plt.show()
```

## 3D Plots

### 3D Scatter Plot

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 5, 1]
z = [10, 15, 20, 25, 30]

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(x, y, z, c='blue', s=100)

ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
plt.title('3D Scatter Plot')
ax.view_init(elev=20, azim=45)
plt.grid(True)
plt.show()
```

### 3D Line Plot

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 5, 1]
z = [4, 5, 3, 2, 6]

plt.figure(figsize=(6, 5))
plot_axis = plt.axes(projection='3d')
plot_axis.plot3D(x, y, z)

plt.title('3D Line Plot')
plt.tight_layout()
plt.show()
```

### 3D Surface Plot

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

x = np.linspace(0, 5, 10)
y = np.linspace(0, 5, 10)
X, Y = np.meshgrid(x, y)

def f(x, y):
    return x**2 + y**2

Z = f(X, Y)

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis', linewidth=0, antialiased=True)

ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
plt.title('3D Surface Plot')
ax.view_init(elev=20, azim=45)
plt.show()
```
