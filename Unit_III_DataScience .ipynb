### Matplotlib & Visualization Concepts

## 1. Introduction to Matplotlib

```python
import matplotlib
import matplotlib.pyplot as plt

print('Matplotlib version:', matplotlib.__version__)

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
rainfall = [120, 85, 60, 45, 90, 140]

plt.plot(months, rainfall, color='steelblue', linewidth=2, marker='D')
plt.title('Monthly Rainfall')
plt.xlabel('Month')
plt.ylabel('mm')
plt.tight_layout()
plt.show()
```

## 2. Subplots – 1D Array (Side by Side)

```python
import matplotlib.pyplot as plt
import numpy as np

hours = np.arange(1, 8)
study_score = [55, 60, 68, 75, 80, 85, 90]
sleep_hours = [8, 7.5, 7, 6.5, 6, 5.5, 5]

plt.subplot(1, 2, 1)
plt.bar(hours, study_score, color='coral')
plt.title('Study vs Score')
plt.xlabel('Study Hours')
plt.ylabel('Score')

plt.subplot(1, 2, 2)
plt.plot(hours, sleep_hours, color='navy', marker='o')
plt.title('Study vs Sleep')
plt.xlabel('Study Hours')
plt.ylabel('Sleep Hours')

plt.suptitle('Student Habits Analysis')
plt.tight_layout()
plt.show()
```

## 3. Subplots – 1D Array (Vertical Stack)

```python
import matplotlib.pyplot as plt
import numpy as np

t = np.linspace(0, 2 * np.pi, 100)

plt.subplot(2, 1, 1)
plt.plot(t, np.sin(t), color='crimson')
plt.title('Sine Wave')
plt.ylabel('sin(t)')

plt.subplot(2, 1, 2)
plt.plot(t, np.cos(t), color='teal')
plt.title('Cosine Wave')
plt.xlabel('t (radians)')
plt.ylabel('cos(t)')

plt.tight_layout()
plt.show()
```

## 4. Subplots – 2D Grid (Multiple Charts)

```python
import matplotlib.pyplot as plt
import numpy as np

products = ['Apples', 'Bananas', 'Cherries']
q1 = [300, 450, 200]
q2 = [350, 400, 250]
q3 = [280, 500, 310]

data_sets = [
    (q1, 'Q1 Sales', 'gold'),
    (q2, 'Q2 Sales', 'skyblue'),
    (q3, 'Q3 Sales', 'salmon'),
    (np.array(q1) + np.array(q2) + np.array(q3), 'Total Sales', 'mediumseagreen')
]

fig, axes = plt.subplots(2, 2, figsize=(10, 7))
axes = axes.flatten()

for i, (vals, title, color) in enumerate(data_sets):
    axes[i].bar(products, vals, color=color)
    axes[i].set_title(title)
    axes[i].set_ylabel('Units')

plt.suptitle('Fruit Store – Quarterly Sales', fontsize=13)
plt.tight_layout()
plt.show()
```

## 5. Multiple Plots using plt.subplots()

```python
import matplotlib.pyplot as plt
import numpy as np

ages = np.array([22, 25, 28, 31, 35, 40, 45, 50, 55, 60])
salary = np.array([30, 38, 50, 60, 75, 85, 90, 95, 88, 80])
expenses = np.array([18, 22, 30, 38, 48, 55, 60, 65, 62, 58])

fig, axs = plt.subplots(2, 2, figsize=(10, 7))

axs[0, 0].plot(ages, salary, 'g-o')
axs[0, 0].set_title('Salary vs Age')

axs[0, 1].scatter(ages, expenses, c='purple', s=60)
axs[0, 1].set_title('Expenses vs Age')

axs[1, 0].bar(ages, salary - expenses, color='tomato')
axs[1, 0].set_title('Savings vs Age')

axs[1, 1].hist(salary, bins=5, color='steelblue', edgecolor='black')
axs[1, 1].set_title('Salary Distribution')

for ax in axs.flat:
    ax.set_xlabel('Age / Value')

plt.suptitle('Personal Finance Overview')
plt.tight_layout()
plt.show()
```

## 6. Controlling Axes (xlim / ylim)

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-10, 10, 300)
y = x ** 2 - 3 * x + 2

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

ax1.plot(x, y, color='darkorange')
ax1.set_title('Full Range')
ax1.set_xlabel('x')
ax1.set_ylabel('f(x)')

ax2.plot(x, y, color='darkorange')
ax2.set_xlim(-3, 5)
ax2.set_ylim(-5, 20)
ax2.set_title('Zoomed View')
ax2.set_xlabel('x')
ax2.set_ylabel('f(x)')

plt.suptitle('Effect of xlim/ylim on Parabola')
plt.tight_layout()
plt.show()
```

## 7. Controlling Ticks (Location & Labels)

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5, 6, 7]
temp_c = [22, 25, 19, 28, 30, 27, 23]

plt.plot(days, temp_c, marker='s', color='tomato', linewidth=2)
plt.xticks([1, 2, 3, 4, 5, 6, 7],
           ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'])
plt.yticks([19, 22, 25, 28, 30],
           ['Cool', 'Mild', 'Warm', 'Hot', 'Very Hot'])
plt.title('Week Temperature with Named Ticks')
plt.tight_layout()
plt.show()
```

## 8. Controlling Ticks (Rotation)

```python
import matplotlib.pyplot as plt

cities = ['Chennai', 'Mumbai', 'Delhi', 'Kolkata', 'Bengaluru', 'Hyderabad']
population = [10.9, 20.7, 32.9, 14.8, 13.2, 10.0]

plt.bar(cities, population, color='mediumpurple', edgecolor='black')
plt.xticks(rotation=40, ha='right')
plt.yticks(rotation=0)
plt.ylabel('Population (millions)')
plt.title('Indian City Populations')
plt.tight_layout()
plt.show()
```

## 9. Controlling Labels (xlabel, ylabel, title)

```python
import matplotlib.pyplot as plt
import numpy as np

speed = np.array([20, 40, 60, 80, 100, 120])
braking_dist = speed ** 2 / 150

plt.plot(speed, braking_dist, 'r^-', linewidth=2)
plt.xlabel('Vehicle Speed (km/h)', fontsize=12, color='darkred')
plt.ylabel('Braking Distance (m)', fontsize=12, color='darkblue')
plt.title('Speed vs Braking Distance', fontsize=14, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

## 10. Controlling Legends

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

plt.plot(x, np.exp(-0.3 * x) * np.sin(x), label='Damped Sine', color='blue', linewidth=2)
plt.plot(x, np.exp(-0.3 * x), label='Decay Envelope', color='orange', linestyle='--')
plt.plot(x, -np.exp(-0.3 * x), label='Negative Envelope', color='orange', linestyle=':')

plt.legend(loc='upper right', fontsize=10, framealpha=0.8)
plt.title('Damped Oscillation with Envelopes')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.tight_layout()
plt.show()
```

## 11. Annotations – Simple Text

```python
import matplotlib.pyplot as plt
import numpy as np

years = np.arange(2018, 2024)
revenue = [12, 18, 15, 25, 30, 28]
peak_idx = revenue.index(max(revenue))

plt.plot(years, revenue, marker='o', color='seagreen', linewidth=2)
plt.annotate('Peak Revenue!',
             xy=(years[peak_idx], revenue[peak_idx]),
             xytext=(years[peak_idx] - 1.5, revenue[peak_idx] + 1.5),
             fontsize=11, color='darkred', fontweight='bold')
plt.title('Company Revenue (2018–2023)')
plt.xlabel('Year')
plt.ylabel('Revenue (Crores)')
plt.tight_layout()
plt.show()
```

## 12. Annotations – With Arrow

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 4 * np.pi, 200)
y = np.sin(x)

plt.plot(x, y, color='royalblue', linewidth=2)

max_x = x[np.argmax(y)]
min_x = x[np.argmin(y)]

plt.annotate('Max',
             xy=(max_x, 1), xytext=(max_x + 1.5, 1.2),
             arrowprops=dict(arrowstyle='->', color='green'),
             color='green', fontsize=10)

plt.annotate('Min',
             xy=(min_x, -1), xytext=(min_x + 1.5, -1.2),
             arrowprops=dict(arrowstyle='->', color='red'),
             color='red', fontsize=10)

plt.title('Sine Wave with Arrow Annotations')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.tight_layout()
plt.show()
```

## 13. Annotations – Different Arrow Styles

```python
import matplotlib.pyplot as plt
import numpy as np

t = np.linspace(0, 2, 50)
v = 9.8 * t

plt.plot(t, v, color='darkcyan', linewidth=2, marker='.')

plt.annotate('Start (v=0)',
             xy=(0, 0), xytext=(0.3, -2),
             arrowprops=dict(arrowstyle='-[', color='navy'),
             color='navy', fontsize=10)

plt.annotate('End (v~19.6)',
             xy=(2, 19.6), xytext=(1.2, 22),
             arrowprops=dict(arrowstyle='|-|', color='purple'),
             color='purple', fontsize=10)

plt.title('Free Fall – Velocity vs Time')
plt.xlabel('Time (s)')
plt.ylabel('Velocity (m/s)')
plt.tight_layout()
plt.show()
```

## 14. Annotations – Bounding Box (bbox)

```python
import matplotlib.pyplot as plt
import numpy as np

categories = ['Python', 'Java', 'C++', 'JavaScript', 'Rust']
popularity = [35, 20, 15, 25, 5]

plt.bar(categories, popularity, color=['#4CAF50', '#2196F3', '#FF5722', '#FFC107', '#9C27B0'])

top_idx = popularity.index(max(popularity))
plt.annotate('Most Popular!',
             xy=(top_idx, popularity[top_idx]),
             xytext=(top_idx + 1, popularity[top_idx] + 2),
             bbox=dict(boxstyle='round,pad=0.4', edgecolor='black', facecolor='lightyellow'),
             arrowprops=dict(facecolor='black', arrowstyle='->'),
             fontsize=10)

plt.title('Programming Language Popularity (%)')
plt.ylabel('%')
plt.tight_layout()
plt.show()
```

## 15. Saving Plots (savefig)

```python
import matplotlib.pyplot as plt
import numpy as np

subjects = ['Maths', 'Physics', 'Chemistry', 'Biology', 'CS']
marks = [88, 75, 92, 65, 98]

plt.figure(figsize=(8, 4))
plt.barh(subjects, marks, color='mediumslateblue', edgecolor='black')
plt.xlabel('Marks')
plt.title('Student Subject Marks')
plt.xlim(0, 110)

for i, v in enumerate(marks):
    plt.text(v + 1, i, str(v), va='center', fontsize=10)

plt.tight_layout()
plt.savefig('student_marks.png', dpi=150, bbox_inches='tight')
plt.show()
print('Plot saved as student_marks.png')
```
