## 1. Seaborn Library

Seaborn is built on top of Matplotlib and makes statistical plotting much easier. It has nicer default styles and works really well with Pandas DataFrames.

### 1.1 Statistical Graphics – Scatter Plot with Grouping (Hue)

Here we use the `hue` parameter to color data points by category. This is one of Seaborn's most useful features.

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# Simple dataset with two groups
data = {
    'x': [1, 2, 3, 4, 5],
    'y': [2, 4, 3, 5, 1],
    'group': ['X', 'Y', 'X', 'Y', 'X']
}

df = pd.DataFrame(data)

# The hue parameter automatically colors points based on the 'group' column
sns.scatterplot(x='x', y='y', hue='group', data=df)

plt.title('Scatter Plot with Group-Based Hue')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

### 1.2 Joint Plot

A joint plot shows the relationship between two variables AND their individual distributions at the same time. Very handy for a quick EDA.

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'x': [1, 2, 3, 4, 5],
    'y': [2, 5, 3, 8, 10]
}

# kind='scatter' shows a scatter plot in the center with histograms on the edges
sns.jointplot(x='x', y='y', kind='scatter', data=data)

plt.show()
```

### 1.3 Pair Plot

A pair plot creates a grid of scatter plots for every pair of columns in your dataset. Diagonal cells show the distribution of each variable. Great for spotting correlations quickly.

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'x': [1, 2, 3, 4, 5],
    'y': [2, 4, 3, 5, 1],
    'z': [10, 15, 20, 25, 30]
}

df = pd.DataFrame(data)

# pairplot automatically creates scatter plots for all column combinations
sns.pairplot(df)
plt.show()
```

### 1.4 Grid Plot using GridSpec

`GridSpec` lets you create a custom grid layout where subplots can span multiple rows or columns — more flexible than the regular `subplots()`.

```python
import matplotlib.pyplot as plt

# Create a 2x3 grid with spacing between plots
Grid_plot = plt.GridSpec(2, 3, wspace=0.8, hspace=0.6)

# Top-left: occupies one cell
plt.subplot(Grid_plot[0, 0])
plt.title('Top Left')
plt.grid(True)

# Top: occupies columns 1 and 2
plt.subplot(Grid_plot[0, 1:])
plt.title('Top Right (spans 2 cols)')
plt.grid(True)

# Bottom: occupies columns 0 and 1
plt.subplot(Grid_plot[1, :2])
plt.title('Bottom Left (spans 2 cols)')
plt.grid(True)

# Bottom-right: occupies one cell
plt.subplot(Grid_plot[1, 2])
plt.title('Bottom Right')
plt.grid(True)

plt.suptitle('Custom Grid Layout using GridSpec', y=1.02)
plt.show()
```

### 1.5 Subplots Grid

This creates a 3×4 grid of empty subplots with a shared title. Useful as a template when you want to fill in plots later.

```python
import matplotlib.pyplot as plt

# Creates a 3x4 grid — 12 subplots total
figure, axes = plt.subplots(3, 4, figsize=(10, 10))

figure.suptitle('Grid Plot using Subplots')
plt.tight_layout()
plt.show()
```

## 2. Multiple Plots

Here we look at different approaches for combining multiple visualizations.

### 2.1 FacetGrid – Column-Based

FacetGrid splits your data by a categorical column and draws the same plot for each group side by side. Very useful for comparisons.

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'x': [1, 2, 3, 4, 5],
    'y': [2, 4, 3, 5, 1],
    'hue': ['A', 'B', 'A', 'B', 'A']
}

df = pd.DataFrame(data)

# col='hue' means one subplot column per unique value in 'hue'
g = sns.FacetGrid(df, col='hue', height=4)
g.map(sns.scatterplot, 'x', 'y')

plt.show()
```

### 2.2 FacetGrid – Row-Based

Same idea, but this time the subplots are stacked in rows instead of side by side.

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'x': [1, 2, 3, 4, 5],
    'y': [2, 4, 3, 5, 1],
    'hue': ['A', 'B', 'A', 'B', 'A']
}

df = pd.DataFrame(data)

# row='hue' stacks the subplots vertically instead of horizontally
g = sns.FacetGrid(df, row='hue', height=4)
g.map(sns.scatterplot, 'x', 'y')

plt.show()
```

### 2.3 Joint Plot (Revisited under Multiple Plots)

Same joint plot example — included here again as it belongs in the Multiple Plots section of the slides.

## 3. Scatter Plot, Line Plot, and Histogram

### 3.1 Scatter Plot

A scatter plot shows the relationship between two numeric variables using dots. Each dot is one data point.

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'x_values': [1, 2, 3, 4, 5],
    'y_values': [2, 5, 3, 8, 10]
}

sns.scatterplot(x='x_values', y='y_values', data=data)

plt.title('Scatterplot Example')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

### 3.2 Line Plot

A line plot connects data points with a line — good for showing trends, especially over time.

```python
import seaborn as sns
import matplotlib.pyplot as plt

x_values = [1, 2, 3, 4, 5]
y_values = [2, 5, 3, 8, 10]

sns.lineplot(x=x_values, y=y_values)

plt.title('Sample Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

### 3.3 Histogram

A histogram shows how data is distributed by grouping values into bins and showing how many values fall into each bin.

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate 1000 random values from a standard normal distribution
data = np.random.randn(1000)

# density=True normalizes the histogram so the area sums to 1
plt.hist(data, density=True)

plt.title('Histogram Example')
plt.xlabel('Value')
plt.ylabel('Density')
plt.show()
```

## 4. Box Plot

Box plots summarize data distribution using five numbers: minimum, Q1 (25th percentile), median, Q3 (75th percentile), and maximum. They're also great for spotting outliers.

### 4.1 Basic Box Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'A': [1, 2, 3, 4, 5],
    'B': [2, 4, 6, 8, 10],
    'C': [3, 6, 9, 12, 15]
}

sns.boxplot(data=data)

plt.title('Box Plot Example')
plt.xlabel('Columns')
plt.ylabel('Values')
plt.show()
```

### 4.2 Box Plot with Outliers

When `whis=1.5` (the default), any point more than 1.5 × IQR away from the box is treated as an outlier and shown as a dot.

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# 30 and 28 are intentional outliers in columns A and B
data = {
    'A': [1, 2, 3, 4, 5, 30],
    'B': [2, 4, 6, 8, 7, 28],
    'C': [3, 6, 9, 5, 2, 7]
}

df = pd.DataFrame(data)

# showmeans=True adds a marker for the mean; whis=1.5 sets the outlier threshold
sns.boxplot(data=df, showmeans=True, whis=1.5)

plt.title('Box Plot with Outliers')
plt.xlabel('Columns')
plt.ylabel('Values')
plt.show()
```

## 5. Playing with Text

Seaborn and Matplotlib can both work with text visualizations — from word clouds to annotated scatter plots.

### 5.1 Word Cloud

A word cloud visualizes word frequency — more frequent words appear larger. You'll need the `wordcloud` library (already installed above).

```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud

text = "This is a sample text for creating a word cloud. Word clouds are a great way to visualize the frequency of words in a text."

# Configure the word cloud appearance
wordcloud = WordCloud(width=800, height=400, background_color='white')
wordcloud.generate(text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')   # No axes needed for a word cloud
plt.show()
```

### 5.2 Text Annotation on a Scatter Plot

You can label each data point with its coordinates using `plt.text()` inside a loop.

```python
import seaborn as sns
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 5, 1]

sns.scatterplot(x=x, y=y)

# Add coordinate labels above each point
for i in range(len(x)):
    plt.text(x[i], y[i], f'({x[i]}, {y[i]})', ha='center', va='bottom')

plt.title('Scatter Plot with Text Annotations')
plt.show()
```

### 5.3 Text as a Plot Element

You can place text anywhere on a plot using `ax.transAxes` coordinates (0 to 1) so it doesn't depend on the data scale.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

# (0.5, 0.5) in transAxes means the dead center of the plot
ax.text(
    0.5, 0.5,
    'This is a sample text on the plot',
    horizontalalignment='center',
    verticalalignment='center',
    transform=ax.transAxes
)

plt.title('Text as a Plot Element')
plt.show()
```

## 6. 3D Plots

For 3D plots, we use `mpl_toolkits.mplot3d` which is part of Matplotlib. Seaborn doesn't natively support 3D, so we fall back to Matplotlib directly.

### 6.1 3D Scatter Plot

Similar to a 2D scatter plot but with a Z-axis. Useful for visualizing three variables at once.

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 5, 1]
z = [10, 15, 20, 25, 30]

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# c sets color, s sets marker size
scatter = ax.scatter(x, y, z, c='blue', s=100)

ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
plt.title('3D Scatter Plot')

# elev = camera elevation angle, azim = camera rotation angle
ax.view_init(elev=20, azim=45)
plt.grid(True)
plt.show()
```

### 6.2 3D Line Plot

Connects data points in 3D space with a line. Useful for trajectories or 3D paths.

```python
import matplotlib.pyplot as plt
import numpy as np

x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 5, 1]
z = [4, 5, 3, 2, 6]

plt.figure(figsize=(6, 5))
plot_axis = plt.axes(projection='3d')

# plot3D draws a line through the 3D points
plot_axis.plot3D(x, y, z)

plt.title('3D Line Plot')
plt.tight_layout()
plt.show()
```

### 6.3 3D Surface Plot

A surface plot visualizes a function of two variables (z = f(x, y)) as a 3D surface. We need to create a 2D mesh grid first, then evaluate z at every (x, y) point.

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

# Create evenly spaced values along both axes
x = np.linspace(0, 5, 10)
y = np.linspace(0, 5, 10)

# meshgrid turns 1D arrays into a 2D grid so we can compute z for every (x, y) pair
X, Y = np.meshgrid(x, y)

# Our function: z = x^2 + y^2 (a paraboloid)
def f(x, y):
    return x**2 + y**2

Z = f(X, Y)

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# cmap sets the color scheme; viridis goes from purple to yellow
surf = ax.plot_surface(X, Y, Z, cmap='viridis', linewidth=0, antialiased=True)

ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
plt.title('3D Surface Plot')

ax.view_init(elev=20, azim=45)
plt.show()
```
